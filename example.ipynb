```python
import os
os.environ['CUDA_VISIBLE_DEVICES'] = '5' 
from OmniGen import OmniGenPipeline
pipe = OmniGenPipeline.from_pretrained("/home/zhouchao/data/OmniGen-v1")
```

```python
layers = {
    "all" : ",".join([str(x) for x in list(range(8,32))]),
}
os.environ["SCALE_LAYER"] = layers["all"]
os.environ["MASK_TYPE"] = "step"
```

```python

instruction = "Make the bike rusty, and add a graffti on the wall, and make the weather rainy."
prompt = f"<img><|image_1|></img>, {instruction}"
print(instruction)
input_image = ["image_0.png"]
    
```

# wo SaaS

```python
import os
os.environ["BEGIN_STEP"] = "0"
os.environ["THRE_STEP"] = "0"
os.environ["SCALE"] = str(1)
os.environ["MASK_THRESH"] = "auto"
images = pipe(
    prompt=prompt, 
    input_images=input_image,  
    height=512, 
    width=512,
    separate_cfg_infer=False,  
    guidance_scale=2.5, 
    img_guidance_scale=1.6,
    seed = 42,
    use_kv_cache=True,)
images[0]
```
